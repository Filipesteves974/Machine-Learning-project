**Abstract**

The following project is meant to answer Cars4You's challenge, a used cars retailer intending to automize price estimation based on their database. The objective is to build a regression able to predict the price with the lowest average absolute error (lower than 10% of the car's value). 

We started by dividing the dataset into training dataset and validation dataset (70/30), correcting typos and data types, paintQuality% was removed, negative values, on certain variables, were eliminated from the dataset by replacing it with NaN, since they do not match factual vehicle information. When it comes to data understanding we wanted to visualize the data with histograms and barplots for the individual analysis, for the multivariate analysis we took a look at the correlations between features, 1 on 1 correlations and distributions between price and the different car brands. We decided not to remove outliers, as information loss would be larger than the benefits of regularized database. Procceding with preprocessing, we created new variables: 'car_age', 'mileage_per_year', 'is_eco', 'is_luxury', 'mileage_bin', and 'tax_to_engine_ratio'.


Before moving forward with all features (strings and floats) were scaled with RobustScaler, and encoded using OneHotEncoder, as well as TargetEncoder. For feature selection the following models were used: Chi-squared (for categorical features), spearman correlation, RFE, Lasso and Regression Tree Feature Importance. These models were crutial in order to find the best features to implement towards our next predictive models - leaving out the selected features: 'has_reportedDamage', 'mileage_per_year' (highly correlated to 'year'), 'car_age', 'previousOwners' and 'tax', because they were selected as not so relevant on two or more steps of selection.


For the moment of truth, the selected predictive models were Linear regression, Ridge, Elastic Net, KNN, Random Forest, HistGradient Boosting and MLP, tested through MAE, MAPE and RMSE. The best performances came from the HistGradientBoostingRegressor with a log target showing a MAE = , MAPE = and R² = in the validation dataset, surpassing the RandomForest model showing a MAE = , MAPE = and R² = in the validation dataset. These evaluation metrics were chosen because they capture different dimensions of forecast error and, together, provide a more complete view of model performance. MAE measures the average error in monetary units, is easy to interpret, and is not very sensitive to extreme values, making it suitable for evaluating typical price forecast error. RMSE, by penalising large errors more heavily, allows us to identify models that make severe mistakes in specific observations, which is especially relevant in a pricing context where large deviations are undesirable. MAPE, on the other hand, expresses error in percentage terms, facilitating relative comparison of performance between models and between different price levels. Together with MedAE (robust to outliers that we did not treat) and R2 (generally used for baseline), these metrics ensure a balanced assessment between interpretability, sensitivity to large errors, and relative comparability, avoiding biased conclusions based on a single performance measure.


Therefore, we recommend the company Cars4You use the Gradient Boosting model as it is the most consistent model to predict car prices with the lowest average absolute error, within the available dataset.

**Regression Benchmarking**

The benchmarking phase was crucial for determining the most effective machine learning algorithm for predicting the car price after the dataset from Cars4You was cleaned and preprocessed to handle missing values, encode categorical variables and scale numerical features, when necessary. This process involved selecting and comparing a diverse range of regressor models across three categories: linear models, non-linear models and  ensemble methods.

*Linear Models:* These included Linear Regression, Ridge Regression, Support Vector Machines and ElasticNet. These models serve as a strong baseline, however they demonstrated the worst results out of every other model, with a Mean Squared Error (MAE) excedding £2000. These results showed no overfitting as well as no underfitting, but we're focused on minimizing the MAE as much as possible

*Non-Linear and Instance-Based Models:* K-Nearest Neighbors Regressor (KNR) was also tested to explore non-linear relationships in the data. KNN showed the biggest overfitting case, as we recommend Cars4You to not use it.

*Ensemble Models (Top Performers!):* Random Forest Regressor and **Gradient Boosting Regressor** were also benchmarked. Among the tested approaches, the ensemble models demonstrated a much improved performance, indicating their ability to capture non-linear relationships and interactions between the vehicle features. 

**Model Optimization**

Our feature engineering helped create relevant features (tax_to_engine_ratio, is_eco and is_luxury) that would improve our models performance by taking them into account and feature selection techniques were explored as well to reduce noise, such as removing low-importance variables and adjusting feature scaling strategies. 
After identifying the strongest models, we moved forward to optimization with the objective of improving our predictive performance. Hyperparameter tuning was performed using mostly grid search to identify optimal parameter configurations, particularly for the Random Forest model and Gradient Boosting, along with experimenting with different values in each parameter (within range, maximum and minimum).
