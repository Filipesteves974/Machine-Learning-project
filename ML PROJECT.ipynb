**Abstract**

The following project is meant to answer Cars4You's challenge, a used cars retailer intending to automize price estimation based on their database from 2020. The objective is to build a regression able to predict the price with the lowest average absolute error (lower than 10% of the car's value). 

We started by dividing the dataset into training dataset and validation dataset (70/30), correcting typos and data types, paintQuality% was removed, NaN values were eliminated from the dataset by only considering positive values as it doesn't make sense to have negative values. When it comes to data understanding we wanted to visualize the data with histograms and barplots for the individual analysis, for the multivariates analysis we took a look at the correlations between features, 1 on 1 correlations and distributions between price and the car brands. WE finish by removing outliers using Interquartile Range. Procceding with preprocessing, we created new variables: 'car_age', 'mileage_per_year', 'is_eco', 'is_luxury', 'mileage_bin', and 'tax_to_engine_ratio'.


Before moving forward with all features (strings and floats) were scalled with RobustScaler, and encoded using OneHotEncoder as well as TargetEncoder. For feature engineering the following models: Chi-squared (for categorical features), spearman correlation, RFE, Lasso and Regression Tree Feature Importance. These models were crutial in order to find the best features to implement towards our next predictive models - leaving out the selected features: year', 'mileage', 'mpg', 'engineSize', 'car_age', 'is_eco', 'is_luxury', 'tax_to_engine_ratio', 'transmission_automatic','transmission_manual', 'transmission_other', 'transmission_semiauto','fuelType_diesel', 'fuelType_electric', 'fuelType_hybrid','fuelType_other', 'fuelType_petrol', 'mileage_bin_0-10k','mileage_bin_10k-50k', 'mileage_bin_50k-100k', 'Brand_Encoded','model_Encoded' (??????)

**Regression Benchmarking**

The benchmarking phase was crucial for determining the most effective machine learning algorithm for predicting the car price after the dataset from Cars4You was cleaned and preprocessed to handle missing values, encode categorical variables and scale numerical features, when necessary. This process involved selecting and comparing a diverse range of regressor models across three categories: linear models, non-linear models and  ensemble methods.

*Linear Models:* These included Linear Regression, Ridge Regression, Support Vector Machines and ElasticNet. These models serve as a strong baseline, however they demonstrated the worst results out of every other model, with a Mean Squared Error (MAE) excedding £2500. These results showed no overfitting as well as no underfitting, but we're focused on minimizing the MAE as much as possible

*Non-Linear and Instance-Based Models:* K-Nearest Neighbors Regressor (KNR) was also tested to explore non-linear relationships in the data. KNN showed the biggest overfitting case, as we recommend Cars4You to not use it.

*Ensemble Models (Top Performers!):* Random Forest Regressor and Gradient Boosting Regressor were also benchmarked. Among the tested approaches, the ensemble models demonstrated a much improved performance, indicating their ability to capture non-linear relationships and interactions between the vehicle features. **Model Optimization**

Our feature engineering helped create relevant features (car_age and is_luxury) that would improve our models performance by taking them into account and feature selection techniques were explored as well to reduce noise, such as removing low-importance variables and adjusting feature scaling strategies. 
After identifying the strongest models, we moved forward to optimization with the objective of improving our predictive performance. Hyperparameter tuning was performed using mostly grid search to identify optimal parameter configurations, particularly for the Random Forest model and Gradient Boosting, along with experimenting with different values in each parameter (within range, maximum and minimum).
